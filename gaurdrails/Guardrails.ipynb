```python
import re

def deterministic_guardil(text : str) -> bool:
    banned_keywords = ["hack", "expliot", "malware", "bomb"]
    return any(kw in text.lower() for kw in banned_keywords)

text_inputs = [
    "How do I hack into database?",
    "What is the capital of france?",
    "Explain how malware spreads?"
]

for inp in text_inputs:
    blocked = deterministic_guardil(inp)
    status = "🚫 BLOCKED" if blocked else "✅ ALLOWED"
    print(f"{status} : {inp}")
```

```python
from langchain_openai import ChatOpenAI

def model_based_gaurdrail(text: str) -> str:
    model = ChatOpenAI(model="gpt-4o-mini", temperature=0)
    prompt = f"""Classify the user input as SAFE or UNSAFE.

UNSAFE means the user asks for direct harmful instructions, illegal access, hacking steps, malware creation, or attack guidance.
SAFE means general knowledge, education, definitions, or harmless questions.

Reply with only SAFE or UNSAFE.

Input: {text}"""
    result = model.invoke([{"role": "user", "content": prompt}])
    return result.content.strip().upper()
```

```python
text_inputs = [
    "How do I hack into database?",
    "What is the capital of france?",
    "Explain how malware spreads?"
]

for inp in text_inputs:
    result = model_based_gaurdrail(inp)
    status = "🚫 UNSAFE" if result == "UNSAFE" else "✅ SAFE"
    print(f"{status}: {inp}")
```
